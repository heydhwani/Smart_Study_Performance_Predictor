```python
df["parental_education_level"] = df["parental_education_level"].fillna(
    df["parental_education_level"].mode()[0]
)
```

```python
df = df.drop(columns=["student_id"])
```

```python
def performance_category(score):
    if score >= 85:
        return "Excellent"
    elif score >= 70:
        return "Good"
    elif score >= 50:
        return "Average"
    else:
        return "Poor"

df["performance"] = df["exam_score"].apply(performance_category)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()

categorical_columns = df.select_dtypes(include=["object", "string"]).columns

for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])
```

```python
df.info()
```

# Data Splitting and Feature Scaling

Before training the machine learning models, the dataset is divided into training and testing sets.

- **Training Set (80%)**: Used to train the models.
- **Testing Set (20%)**: Used to evaluate model performance.
- **StandardScaler** is applied to normalize the feature values.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

## Preparing Data for Regression

For regression, the target variable is **exam_score**.

```python
X_reg = df.drop(columns=["exam_score", "performance"])
y_reg = df["exam_score"]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)
```

## Preparing Data for Classification

For classification, the target variable is **performance**.

```python
X_clf = df.drop(columns=["exam_score", "performance"])
y_clf = df["performance"]

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf,
    y_clf,
    test_size=0.2,
    random_state=42
)
```

## Feature Scaling

Feature scaling is performed using **StandardScaler** to ensure that numerical features have a similar scale. This helps improve the performance of machine learning algorithms.

```python
scaler_reg = StandardScaler()

X_train_reg = scaler_reg.fit_transform(X_train_reg)
X_test_reg = scaler_reg.transform(X_test_reg)

scaler_clf = StandardScaler()

X_train_clf = scaler_clf.fit_transform(X_train_clf)
X_test_clf = scaler_clf.transform(X_test_clf)
```

## Verifying Dataset Shapes

The following code checks the dimensions of the training and testing datasets for both regression and classification tasks.

```python
print("Regression Training Shape :", X_train_reg.shape)
print("Regression Testing Shape  :", X_test_reg.shape)

print()

print("Classification Training Shape :", X_train_clf.shape)
print("Classification Testing Shape  :", X_test_clf.shape)
```

# Regression Model

In this section, a **Random Forest Regressor** is trained to predict students' exam scores.

Random Forest is an ensemble learning algorithm that combines multiple decision trees to improve prediction accuracy and reduce overfitting.

```python
from sklearn.ensemble import RandomForestRegressor
```

## Model Training

The regression model is trained using the training dataset.

```python
regressor = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

regressor.fit(X_train_reg, y_train_reg)
```

## Model Prediction

The trained model is used to predict exam scores for the testing dataset.

```python
y_pred_reg = regressor.predict(X_test_reg)
```

## Regression Model Evaluation

The performance of the regression model is evaluated using:

- R² Score
- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)

```python
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    root_mean_squared_error
)

r2 = r2_score(y_test_reg, y_pred_reg)
mae = mean_absolute_error(y_test_reg, y_pred_reg)
rmse = root_mean_squared_error(y_test_reg, y_pred_reg)

print(f"R² Score : {r2:.4f}")
print(f"MAE      : {mae:.4f}")
print(f"RMSE     : {rmse:.4f}")
```

## Actual vs Predicted Exam Scores

This visualization compares the actual exam scores with the predicted scores. A good regression model should produce points close to the diagonal line.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(y_test_reg, y_pred_reg, alpha=0.7)

plt.plot(
    [y_test_reg.min(), y_test_reg.max()],
    [y_test_reg.min(), y_test_reg.max()],
    color="red",
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")
plt.title("Actual vs Predicted Exam Scores")

plt.grid(alpha=0.3)

plt.show()
```

### Observation

The scatter plot shows that most predicted exam scores lie close to the diagonal reference line.

This indicates that the Random Forest Regressor is able to predict student exam scores with good accuracy. Although a few outliers are present, the overall prediction trend closely matches the actual values, demonstrating strong model performance.
